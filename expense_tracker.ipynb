```python
expenses = []

while True:
    print("==========Expense tracker==========")
    print("1-add Expense")
    print("2-show expenses")
    print("3-show total")
    print("4-search by category")
    print("5-delete expense")
    print("6-exit")
    try:
        choice = int(input("enter your choice (from 1-6): "))
    except ValueError:
        print("invalid choice! please enter a number from 1-6")
        continue
  # 1. add expense
    if choice == 1:
        exp_name = input("please enter expense name :")
        while True:
            try:
                exp_amount = float(input("please enter expense amount :"))
                break
            except ValueError:
                print("invalid ! please enter a number")

        exp_category = input("please enter expense category :")

        expense = {
            "name": exp_name,
            "amount": exp_amount,
            "category": exp_category
        }
        expenses.append(expense)
        print("expense added successfully!")

    # 2.show expenses
    elif choice == 2:
        if not expenses:
            print("No expenses added yet!")
        else:
            print("\n--- All Expenses ---")
            for exp in expenses:
                print(f"Name: {exp['name']} | Amount: {exp['amount']} | Category: {exp['category']}")
            print()
        pass

    # 3.show total
    elif choice == 3:
      if not expenses:
        print("No expenses added yet!")
      else:
        total = sum(exp['amount'] for exp in expenses)
        print(f"Total amount of expenses spent: " , total)
        pass

   # 4.serach by category
    elif choice == 4:
      if not expenses:
        print("No expenses added yet!")
      else:
        category_name = input("please enter expense category :")

        for exp in expenses:
            if exp['category'].lower()==category_name.lower():
                print(f"Name: {exp['name']} | Amount: {exp['amount']} | Category: {exp['category']}")
                print()
        pass
  # 5.delete expenses
    elif choice == 5:
      if not expenses:
        print("no expenses added yet!")
      else:
        del_name = input("please enter expense name to delete :")
        for exp in expenses:
            if exp['name'].lower()==del_name.lower():
                expenses.remove(exp)
                print("expense deleted succesfully!")
                break
        pass

    elif choice == 6:
        print("exiting expense tracker...goodbye!")
        break
```
